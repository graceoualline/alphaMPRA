```python
import json
import importlib.util
import numpy as np
import pandas as pd
import jax
import jax.numpy as jnp
from scipy.stats import pearsonr
from tqdm.auto import tqdm
from alphagenome_ft_mpra.hub import download
from alphagenome.models import dna_model as ag_dna_model
from alphagenome_research.model import dna_model as research_dna_model

CELLS  = ['K562', 'HepG2', 'SKNSH']

gosai = pd.read_table('/home/go274/scratch_pi_skr2/go274/coda_data/DATA-Table_S2__MPRA_dataset.txt',
                      sep='\t', header=0, low_memory=False)
gosai['chr'] = gosai['chr'].astype(str)   # column is mixed int/str ('X')

se_cols    = [f'{c}_lfcSE' for c in CELLS]
label_cols = [f'{c}_log2FC' for c in CELLS]

# chr 7/13, lfcSE < 1 in all three cells, exact 200-mers: same protocol as load_malinois_model.ipynb
test = gosai.loc[
    gosai['chr'].isin(['7', '13'])
    & (gosai[se_cols].max(axis=1) < 1.0)
    & (gosai['sequence'].str.len() == 200)
].reset_index(drop=True)

assert set(test['chr']) == {'7', '13'}, f"unexpected chromosomes: {sorted(set(test['chr']))}"
assert test[label_cols].notna().all().all(), f"NaN labels after filtering: {test[label_cols].isna().sum().to_dict()}"
print(f'test rows: {len(test)}')   # 60055 when I scored Malinois this way
```

```python
import json
import importlib.util
import numpy as np
import pandas as pd
import jax
import jax.numpy as jnp
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
from tqdm.auto import tqdm
from alphagenome_ft_mpra.hub import download
from alphagenome.models import dna_model as ag_dna_model
from alphagenome_research.model import dna_model as research_dna_model

# 200 bp episomal flanks (MPRA_UPSTREAM / MPRA_DOWNSTREAM from boda), read from baseline_CODA/parse_args.py
_spec = importlib.util.spec_from_file_location(
    'coda_parse_args', '/home/go274/scratch_pi_skr2/go274/manual_MPRA_models/baseline_CODA/parse_args.py')
_coda = importlib.util.module_from_spec(_spec); _spec.loader.exec_module(_coda)
FLANK_5 = _coda.MPRA_UPSTREAM[-200:]
FLANK_3 = _coda.MPRA_DOWNSTREAM[:200]
assert len(FLANK_5) == len(FLANK_3) == 200, (len(FLANK_5), len(FLANK_3))

_LUT = np.full(256, -1, dtype=np.int8)
for _i, _b in enumerate('ACGT'):
    _LUT[ord(_b)] = _i

ORGANISM = ag_dna_model.Organism.HOMO_SAPIENS
ORG_IDX  = research_dna_model.convert_to_organism_index(ORGANISM)


class GosaiPredictor:
    """predict_sequences() for the JAX Gosai-<cell>-optimal checkpoints.

    Input: 200 bp inserts. Each is placed between the 200 bp episomal flanks
    to make the 600 bp construct the model was trained on.
    """

    def __init__(self, model, checkpoint_name, stage='stage2'):
        self.model = model
        cfg = json.loads((download(checkpoint_name) / stage / 'config.json').read_text())
        self.head = cfg['custom_heads'][0]
        strand_reindexing = jax.device_put(model._metadata[ORGANISM].strand_reindexing,
                                           model._device_context._device)
        head = self.head

        @jax.jit
        def _f(params, x):
            B, L = x.shape[0], x.shape[1]
            out = model._predict(params, model._state, x,
                                 jnp.full((B,), ORG_IDX, dtype=jnp.int32),
                                 requested_outputs=None,
                                 negative_strand_mask=jnp.zeros((B, L), dtype=jnp.bool_),
                                 strand_reindexing=strand_reindexing)
            return out[head].reshape(B)
        self._f = _f

    @staticmethod
    def build_construct(seqs):
        """200 bp inserts -> (B, 600, 4) one-hot, ACGT order, N = all-zero row."""
        assert all(len(s) == 200 for s in seqs), 'inserts must be exactly 200 bp'
        full = [FLANK_5 + s.upper() + FLANK_3 for s in seqs]
        idx = _LUT[np.frombuffer(''.join(full).encode(), dtype=np.uint8)].reshape(len(full), 600)
        x = np.zeros((len(full), 600, 4), dtype=np.float32)
        rows, pos = np.nonzero(idx >= 0)
        x[rows, pos, idx[rows, pos]] = 1.0
        return x

    def predict_sequences(self, seqs, rc_average=True, batch_size=64, progress=True):
        seqs = list(seqs)
        out = []
        it = range(0, len(seqs), batch_size)
        for i in (tqdm(it, desc=self.head) if progress else it):
            x = jnp.asarray(self.build_construct(seqs[i:i + batch_size]))
            p = np.asarray(self._f(self.model._params, x))
            if rc_average:   # reverse positions and swap A<->T, C<->G
                p = (p + np.asarray(self._f(self.model._params, x[:, ::-1, ::-1]))) / 2
            out.append(p)
        preds = np.concatenate(out)
        assert preds.shape == (len(seqs),), f'{preds.shape} vs {len(seqs)} seqs'
        assert np.isfinite(preds).all(), f'{(~np.isfinite(preds)).sum()} non-finite predictions'
        return preds


predictors = {
    'K562':  GosaiPredictor(model_k562,  'Gosai-K562-optimal'),
    'HepG2': GosaiPredictor(model_hepg2, 'Gosai-HepG2-optimal'),
    'SKNSH': GosaiPredictor(model_sknsh, 'Gosai-SKNSH-optimal'),
}
print({c: p.head for c, p in predictors.items()})
```

```python
def plot_cell(observed, predicted, cell, title=None, save_name=None, ax=None):
    """Density scatter of observed vs predicted log2FC for one cell type."""
    x = np.asarray(observed, dtype=float)
    y = np.asarray(predicted, dtype=float)
    assert x.shape == y.shape, f'observed {x.shape} vs predicted {y.shape}'
    ok = ~(np.isnan(x) | np.isnan(y))
    x, y = x[ok], y[ok]

    own_fig = ax is None
    if own_fig:
        fig, ax = plt.subplots(figsize=(4, 4))

    all_vals = np.concatenate([x, y])
    vmin, vmax = all_vals.min(), all_vals.max()
    margin = (vmax - vmin) * 0.05
    lim = (vmin - margin, vmax + margin)

    sns.histplot(x=x, y=y, ax=ax, bins=50, binrange=(lim, lim),
                 cmap='viridis', cbar=False, pthresh=0.02)
    ax.plot(lim, lim, color='crimson', linewidth=0.9, linestyle='--', zorder=3)
    ax.axhline(0, color='gray', linewidth=0.7, linestyle=':', zorder=2)
    ax.axvline(0, color='gray', linewidth=0.7, linestyle=':', zorder=2)
    ax.set_xlim(lim); ax.set_ylim(lim)

    r, pval = pearsonr(x, y)
    stars = '***' if pval < 0.001 else '**' if pval < 0.01 else '*' if pval < 0.05 else ''
    ax.text(0.05, 0.95, f'r = {r:.3f}{stars}', transform=ax.transAxes, fontsize=7,
            va='top', ha='left',
            bbox=dict(boxstyle='round,pad=0.2', fc='white', ec='none', alpha=0.7))
    ax.text(0.5, 0.97, f'{cell} (n={len(x)})', transform=ax.transAxes,
            ha='center', va='top', fontsize=8, fontweight='bold')
    ax.set_xlabel(f'{cell} observed log2FC', fontsize=8)
    ax.set_ylabel(f'{cell} predicted log2FC', fontsize=8)
    ax.tick_params(labelsize=6)

    if own_fig:
        if title:
            fig.suptitle(title, fontsize=10)
        fig.tight_layout()
        if save_name:
            fig.savefig(save_name, dpi=200, bbox_inches='tight')
        plt.show()
    return r
```

```python
# full test set, one cell type at a time
preds_k562  = predictors['K562'].predict_sequences(test['sequence'])
preds_hepg2 = predictors['HepG2'].predict_sequences(test['sequence'])
preds_sknsh = predictors['SKNSH'].predict_sequences(test['sequence'])
```

```python
# one cell type at a time
plot_cell(test['K562_log2FC'],  preds_k562,  'K562',  title='AlphaGenome Gosai stage2 - chr 7/13',
          save_name='alphampra_K562epi_test.png')
plot_cell(test['HepG2_log2FC'], preds_hepg2, 'HepG2', title='AlphaGenome Gosai stage2 - chr 7/13',
          save_name='alphampra_HepG2epi_test.png')
plot_cell(test['SKNSH_log2FC'], preds_sknsh, 'SKNSH', title='AlphaGenome Gosai stage2 - chr 7/13',
          save_name='alphampra_SKNSHepi_test.png')

# or all three side by side
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
rs = [plot_cell(test[f'{c}_log2FC'], p, c, ax=ax)
      for ax, c, p in zip(axes, CELLS, [preds_k562, preds_hepg2, preds_sknsh])]
fig.suptitle(f'AlphaGenome Gosai stage2 - chr 7/13\nmean Pearson r = {np.mean(rs):.3f}', fontsize=11, y=1.05)
fig.tight_layout()
fig.savefig('alphampra_chr7_13_test.png', dpi=200, bbox_inches='tight')
plt.show()
```
